```python
from pathlib import Path
import pandas as pd

# Find the main FlowCast project folder
PROJECT_DIR = Path.cwd().parent
DATA_DIR = PROJECT_DIR / "data"

print("Project folder:", PROJECT_DIR)
print("Data folder exists:", DATA_DIR.exists())
print("Files in data folder:")

for file in DATA_DIR.iterdir():
    print(file.name)
```

```python
# Load the main traffic dataset
data_file = DATA_DIR / "traffic_sensor_log.csv"

df = pd.read_csv(data_file)

print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
# Check missing values in each column
print("MISSING VALUES")
print(df.isnull().sum())

# Check duplicate traffic records
print("\nDUPLICATE ROWS")
print("Exact duplicate rows:", df.duplicated().sum())

print("\nDuplicate road/time records:")
print(df.duplicated(subset=["road_id", "date", "time"]).sum())

# Check data types
print("\nDATA TYPES")
print(df.dtypes)
```

```python
# Check values that may be physically impossible

print("Negative traffic volume:")
print((df["traffic_volume"] < 0).sum())

print("\nNegative average speed:")
print((df["avg_speed"] < 0).sum())

print("\nAverage speed above 200 km/h:")
print((df["avg_speed"] > 200).sum())

print("\nOccupancy below 0:")
print((df["occupancy"] < 0).sum())

print("\nOccupancy above 100:")
print((df["occupancy"] > 100).sum())

print("\nNegative accident counts:")
print((df["accident_count"] < 0).sum())

print("\nNegative travel times:")
print((df["travel_time"] < 0).sum())
```

```python
print("Congestion level counts:")
print(df["congestion_level"].value_counts(dropna=False))

print("\nUnique congestion labels:")
print(df["congestion_level"].dropna().unique())
```

```python
# Check date range and road coverage

df["date"] = pd.to_datetime(df["date"], errors="coerce")

print("Earliest date:", df["date"].min())
print("Latest date:", df["date"].max())

print("\nNumber of unique roads:", df["road_id"].nunique())

print("\nRecords per road:")
print(df["road_id"].value_counts().sort_index())

print("\nNumber of unique time values:", df["time"].nunique())
print("\nSample time values:")
print(df["time"].dropna().unique()[:10])
```

```python
# Check whether each road has duplicate date/time records
print("Duplicate road-date-time combinations:")
print(
    df.duplicated(
        subset=["road_id", "date", "time"]
    ).sum()
)

# Check how many records each road has per day
records_per_day = df.groupby(
    ["road_id", "date"]
).size()

print("\nRecords per road per day — summary:")
print(records_per_day.describe())

print("\nDays with fewer than 48 records:")
print((records_per_day < 48).sum())

print("\nDays with more than 48 records:")
print((records_per_day > 48).sum())
```

```python
# Find road-days with the fewest records
print("10 road-days with the fewest records:")

print(
    records_per_day
    .sort_values()
    .head(10)
)

print("\n10 road-days with the most records:")

print(
    records_per_day
    .sort_values(ascending=False)
    .head(10)
)
```

```python
# Display examples of duplicate road/date/time records

duplicate_mask = df.duplicated(
    subset=["road_id", "date", "time"],
    keep=False
)

duplicate_examples = (
    df.loc[duplicate_mask]
    .sort_values(["road_id", "date", "time"])
)

print("Total rows involved in duplicate groups:",
      len(duplicate_examples))

display(duplicate_examples.head(10))
```

```python
# Check exact duplicates across all columns

print("Exact duplicate rows:", df.duplicated().sum())

# Check whether duplicate road/time records have conflicting measurements
duplicate_groups = (
    df.groupby(["road_id", "date", "time"], dropna=False)
      .size()
)

print("\nRoad/date/time groups with multiple records:",
      (duplicate_groups > 1).sum())

print("\nNumber of records in duplicate groups:",
      duplicate_groups[duplicate_groups > 1].sum())
```

```python
# Create a working copy of the traffic dataset
df_traffic = df.copy()

rows_before = len(df_traffic)

# Keep one record for each road, date, and time
df_traffic = df_traffic.drop_duplicates(
    subset=["road_id", "date", "time"],
    keep="first"
).copy()

rows_after = len(df_traffic)

print("Rows before cleaning:", rows_before)
print("Rows after removing duplicates:", rows_after)
print("Duplicate records removed:", rows_before - rows_after)

print(
    "Duplicate road/date/time keys remaining:",
    df_traffic.duplicated(
        subset=["road_id", "date", "time"]
    ).sum()
)
```

```python
# Check invalid sensor measurements after removing duplicates

print("Negative traffic volume:",
      (df_traffic["traffic_volume"] < 0).sum())

print("Negative average speed:",
      (df_traffic["avg_speed"] < 0).sum())

print("Average speed above 200 km/h:",
      (df_traffic["avg_speed"] > 200).sum())

print("Occupancy below 0:",
      (df_traffic["occupancy"] < 0).sum())

print("Occupancy above 100:",
      (df_traffic["occupancy"] > 100).sum())

print("Missing traffic volume:",
      df_traffic["traffic_volume"].isna().sum())

print("Missing average speed:",
      df_traffic["avg_speed"].isna().sum())

print("Missing occupancy:",
      df_traffic["occupancy"].isna().sum())
```

```python
# Create a separate cleaned working copy
df_clean = df_traffic.copy()

# Replace impossible sensor readings with missing values
df_clean.loc[df_clean["traffic_volume"] < 0, "traffic_volume"] = float("nan")

df_clean.loc[df_clean["avg_speed"] > 200, "avg_speed"] = float("nan")

df_clean.loc[df_clean["occupancy"] > 100, "occupancy"] = float("nan")

print("Invalid values handled.")
print("\nRemaining negative traffic volume:",
      (df_clean["traffic_volume"] < 0).sum())

print("Remaining speed above 200:",
      (df_clean["avg_speed"] > 200).sum())

print("Remaining occupancy above 100:",
      (df_clean["occupancy"] > 100).sum())

print("\nMissing values after handling invalid readings:")
print(df_clean[["traffic_volume", "avg_speed", "occupancy"]].isna().sum())
```

```python
# Prepare timestamps for time-based cleaning
df_clean["timestamp"] = pd.to_datetime(
    df_clean["date"].astype(str) + " " + df_clean["time"].astype(str),
    errors="coerce"
)

# Sort each road's records in chronological order
df_clean = df_clean.sort_values(
    ["road_id", "timestamp"]
).copy()

# Fill missing sensor measurements using nearby records
sensor_columns = ["traffic_volume", "avg_speed", "occupancy"]

for col in sensor_columns:
    df_clean[col] = df_clean.groupby("road_id")[col].transform(
        lambda s: s.interpolate(limit_direction="both")
    )

# Check remaining missing values
print("Missing sensor values after interpolation:")
print(df_clean[sensor_columns].isna().sum())

print("\nTotal rows:", len(df_clean))
```

```python
# Verify sensor values after cleaning

print("Traffic volume minimum:",
      df_clean["traffic_volume"].min())

print("Average speed minimum:",
      df_clean["avg_speed"].min())

print("Average speed maximum:",
      df_clean["avg_speed"].max())

print("Occupancy minimum:",
      df_clean["occupancy"].min())

print("Occupancy maximum:",
      df_clean["occupancy"].max())

print("\nMissing congestion labels:",
      df_clean["congestion_level"].isna().sum())

print("\nCongestion label counts:")
print(df_clean["congestion_level"].value_counts(dropna=False))
```

```python
# Check road capacity and traffic volume
print("Road capacity summary:")
print(df_clean["road_capacity"].describe())

print("\nTraffic volume summary:")
print(df_clean["traffic_volume"].describe())

print("\nExisting congestion labels:")
print(df_clean["congestion_level"].value_counts())
```

```python
# Calculate the traffic-to-capacity ratio for a 30-minute window
df_clean["capacity_30min"] = df_clean["road_capacity"] / 2

df_clean["volume_capacity_ratio"] = (
    df_clean["traffic_volume"] / df_clean["capacity_30min"]
)

# Inspect the ratio before assigning missing congestion labels
print("Traffic-to-capacity ratio summary:")
display(df_clean["volume_capacity_ratio"].describe())

print("\nMissing congestion labels:")
print(df_clean["congestion_level"].isna().sum())
```

```python
# Check how existing congestion labels relate to the ratio
display(
    df_clean.groupby("congestion_level", dropna=False)[
        "volume_capacity_ratio"
    ].agg(["count", "min", "median", "max"])
)
```

```python
# Fill only missing congestion labels using the Data Dictionary
missing_mask = df_clean["congestion_level"].isna()

df_clean.loc[missing_mask, "congestion_level"] = pd.cut(
    df_clean.loc[missing_mask, "volume_capacity_ratio"],
    bins=[-float("inf"), 0.50, 0.80, 1.00, float("inf")],
    labels=["Free-flow", "Moderate", "Heavy", "Severe"],
    right=False
)

print("Missing labels remaining:",
      df_clean["congestion_level"].isna().sum())

print("\nCongestion label counts:")
print(df_clean["congestion_level"].value_counts())
```

```python
# Create the processed data folder if it doesn't exist
from pathlib import Path

PROCESSED_DIR = PROJECT_DIR / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Save the cleaned traffic dataset
output_path = PROCESSED_DIR / "traffic_cleaned.csv"
df_clean.to_csv(output_path, index=False)

print("Cleaned traffic dataset saved successfully!")
print("File location:", output_path)
print("Total rows:", len(df_clean))
print("Total columns:", len(df_clean.columns))
```

```python
# FLOWCAST - DATA QUALITY REPORT

import pandas as pd
import os

quality_report = pd.DataFrame({
    "Column": df_clean.columns,
    "Data_Type": df_clean.dtypes.astype(str).values,
    "Missing_Values": df_clean.isnull().sum().values,
    "Missing_Percentage": (
        df_clean.isnull().mean().values * 100
    ).round(2)
})

# Save detailed quality report
report_path = "../data/processed/data_quality_report.csv"

quality_report.to_csv(report_path, index=False)

print("DATA QUALITY REPORT")
print("-------------------")
print("Total rows:", len(df_clean))
print("Total columns:", len(df_clean.columns))
print("Duplicate rows:", df_clean.duplicated().sum())
print("Total missing values:", df_clean.isnull().sum().sum())

print("\nColumn-wise report:")
display(quality_report)

print("\nReport saved successfully:", report_path)
```

```python
# FLOWCAST - INTEGRATED DATASET VERIFICATION

import pandas as pd
import os

data_path = "../data/processed/flowcast_features.csv"

if os.path.exists(data_path):
    integrated_df = pd.read_csv(data_path)

    print("INTEGRATED DATASET")
    print("------------------")
    print("Dataset shape:", integrated_df.shape)
    print("\nAvailable columns:")
    print(integrated_df.columns.tolist())

    print("\nRoad ID available:",
          "road_id" in integrated_df.columns)

    print("Timestamp available:",
          "timestamp" in integrated_df.columns)

    weather_cols = [
        c for c in integrated_df.columns
        if any(x in c.lower() for x in
               ["weather", "temperature", "rain", "humidity", "wind"])
    ]

    calendar_cols = [
        c for c in integrated_df.columns
        if any(x in c.lower() for x in
               ["holiday", "event", "weekend", "calendar"])
    ]

    print("\nWeather-related columns:", weather_cols)
    print("Calendar-related columns:", calendar_cols)

else:
    print("Integrated dataset not found:", data_path)
```
