```python
import pandas as pd
from pathlib import Path

# Set the FlowCast project folder
PROJECT_DIR = Path.home() / "Desktop" / "FlowCast"

# Load the original calendar dataset
calendar_path = PROJECT_DIR / "data" / "calendar_events.csv"

df_calendar = pd.read_csv(calendar_path)

# Display the first 5 rows
display(df_calendar.head())

# Check dataset size
print("Dataset shape:", df_calendar.shape)

# Check column names
print("\nColumn names:")
print(df_calendar.columns.tolist())

# Check missing values
print("\nMissing values:")
print(df_calendar.isna().sum())

# Check duplicate rows
print("\nExact duplicate rows:", df_calendar.duplicated().sum())
```

```python
# Make a copy so the original data remains unchanged
df_calendar_clean = df_calendar.copy()

# Convert date to datetime format
df_calendar_clean["date"] = pd.to_datetime(
    df_calendar_clean["date"],
    errors="coerce"
)

# Remove duplicate dates, if any
df_calendar_clean = df_calendar_clean.drop_duplicates(
    subset=["date"],
    keep="first"
).copy()

# Standardize holiday and event names
for col in ["holiday_name", "event_name"]:
    df_calendar_clean[col] = (
        df_calendar_clean[col]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
    )

# Ensure flag columns are numeric
flag_columns = ["public_holiday", "event_flag", "roadwork_flag"]

for col in flag_columns:
    df_calendar_clean[col] = pd.to_numeric(
        df_calendar_clean[col],
        errors="coerce"
    )

# Check the cleaned dataset
print("Dataset shape:", df_calendar_clean.shape)
print("\nMissing values:")
print(df_calendar_clean.isna().sum())

print("\nDuplicate dates remaining:",
      df_calendar_clean["date"].duplicated().sum())

display(df_calendar_clean.head(10))
```

```python
# Create the processed-data folder
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Save the cleaned calendar dataset
output_path = PROCESSED_DIR / "calendar_cleaned.csv"

df_calendar_clean.to_csv(output_path, index=False)

print("Calendar dataset saved successfully!")
print("File location:", output_path)
print("Total rows:", len(df_calendar_clean))
print("Total columns:", len(df_calendar_clean.columns))
```
